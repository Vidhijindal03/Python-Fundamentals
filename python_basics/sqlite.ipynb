```python
import sqlite3
```

```python
## connect to a sqlite database
connection=sqlite3.connect('example.db')
connection
```

```python
cursor=connection.cursor()
```

```python
## create a table
cursor.execute('''
    CREATE TABLE IF NOT EXISTS employees (
        id INTEGER PRIMARY KEY,
        name TEXT NOT NULL,
        age Integer,
        department text
    )
''')

##commit the changes
connection.commit()
```

```python
cursor.execute('''
Select * from employees

''')
```

```python
cursor.execute('''

Insert into employees (name,age,department) values ('Alice',30,'HR')
''')

cursor.execute('''
Insert into employees (name,age,department) values ('Bob',25,'IT')
''')

cursor.execute('''
Insert into employees (name,age,department) values ('Charlie',28,'Finance')         
''')

connection.commit()
```

```python
## Query the data from the table
cursor.execute('''
Select * from employees
''')
rows=cursor.fetchall()

for row in rows:
    print(row)
    
```
